```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
model = YOLO('yolo11n.pt')  
```

```python
# Path to image
img_path = 'img7.jpg'

# Run inference (CPU-only)
results = model(img_path, device='cpu')

# Display with OpenCV or Matplotlib
annotated = results[0].plot()  # Returns annotated frame (numpy array)

plt.figure(figsize=(12, 8))
plt.imshow(cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
import cv2
from ultralytics import YOLO

# Load model
model = YOLO('yolov11.pt')  # Replace with your actual model path

# Open webcam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Resize for better speed (optional)
    # frame = cv2.resize(frame, (640, 480))

    # Run YOLOv11 detection (on CPU)
    results = model(frame, device='cpu')

    # Annotate frame with bounding boxes
    annotated_frame = results[0].plot()

    # Show the live detection
    cv2.imshow("YOLOv11 Live Detection", annotated_frame)

    # Exit loop if 'q' is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources
cap.release()
cv2.destroyAllWindows()
```
